# Cross-linking induced termination sites (CITS) calling using XLCenter

This example runs XLCenter to generate precise CITS centers for RBFOX2 chromosome 1 using ENCODE eCLIP data.

To call CITS (call=1) using a posterior probability threshold of 0.9,

```python
python xlcenter.py callsites \
-i examples/RBFOX2_eCLIP_rep1_chr1.bam \
-o output/RBFOX2_eCLIP_rep1_chr1 \
-f examples/ucsc.hg19.sizes \
-e eCLIP \
-t 0.9 \
-d bum
```

To filter the CITS using a more stringent  posterior probability threshold of 0.999,

```python
python xlcenter.py threshold \
-p output/RBFOX2_eCLIP_rep1_chr1.txt \
-b output/RBFOX2_eCLIP_rep1_chr1_0.999.bed \
-t 0.999 \
-d beta
```
